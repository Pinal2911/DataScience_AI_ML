```python
#password length

def isstrong_psw(psw):
    if(len(psw) < 8):
        return False
    if not any (char.isdigit() for char in psw):
        return False
    if not any(char.islower() for char in psw):
        return False
    if not any(char.isupper() for char in psw):
        return False
    if not any(char in '!@#$%^&*()-_' for char in psw):
        return False
    return True
    
```

```python
print(isstrong_psw("Pinal"))
print(isstrong_psw("pinal@29111"))
print(isstrong_psw("Pinal@2911"))
```

```python
#string plaindrome

def plaindrome(s):
    s=s.lower().replace(" ",".")
    return s==s[::-1]
```

```python
plaindrome("pinal")
```

```python
plaindrome("man nam")
```

```python
#factorial in recusrion

def factorial(n):
    if n==0:
        return 1
    else:
        return n*factorial(n-1)
```

```python
print(factorial(5))
```
