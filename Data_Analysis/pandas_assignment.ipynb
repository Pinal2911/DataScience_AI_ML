```python
'''### Assignment 4: Data Aggregation

1. Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'. Fill the 'Category' column with random categories ('A', 'B', 'C') and the 'Value' column with random integers. Group the DataFrame by 'Category' and compute the sum and mean of 'Value' for each category.'''

data=np.array([
   ('A',10),
   ('B',33),
   ('C',45),
   ('A',65),
   ('B',22),
   ('B',12),
   ('B',56),
   ('C',320) 
],dtype=[('Category','U10'),('Value','i4')])

df=pd.DataFrame(data)
groupedmean=df.groupby('Category')['Value'].mean()
groupedsum=df.groupby('Category')['Value'].sum()
print("mean = ",groupedmean)
print("groupedsum = ",groupedsum)
```

```python
df1 = pd.DataFrame({'A': np.random.randint(1, 100, size=3), 'B': np.random.randint(1, 100, size=3)})
df2 = pd.DataFrame({'C': np.random.randint(1, 100, size=3), 'D': np.random.randint(1, 100, size=3)})
print("DataFrame 1:")
print(df1)
print("DataFrame 2:")
print(df2)

# Concatenate the DataFrames along the rows
concat_rows = pd.concat([df1, df2], axis=0)
print("Concatenated DataFrame (rows):")
print(concat_rows)

# Concatenate the DataFrames along the columns
concat_columns = pd.concat([df1, df2], axis=1)
print("Concatenated DataFrame (columns):")
print(concat_columns)
```

```python
'''### Assignment 6: Time Series Analysis

1. Create a Pandas DataFrame with a datetime index and one column filled with random integers. Resample the DataFrame to compute the monthly mean of the values.'''

date_values=pd.date_range(start='2026-01-01',end='2026-12-31',freq='D')
df=pd.DataFrame(date_values,columns=['date_value'])
df['values']=np.random.randint(1,100,size=len(df['date_value']))
#setting datevalue as index
df.set_index(df['date_value'],inplace=True)
print(df)
df.resample('ME').mean()
```

```python
'''Create a Pandas DataFrame with a datetime index ranging from '2021-01-01' to '2021-12-31' and one column filled with random integers. Compute the rolling mean with a window of 7 days.'''
#rolling mean
date_values=pd.date_range(start='2021-01-01',end='2021-12-31',freq='D')
df=pd.DataFrame(date_values,columns=['date_column'])
df['num']=np.random.randint(0,100,size=len(date_values),dtype=int)

df.set_index('date_column',inplace=True)
df.rolling(window=7).mean()
```
