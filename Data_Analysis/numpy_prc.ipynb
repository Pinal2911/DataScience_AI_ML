```python
#attributes of array
arr=np.array([[1,2,3],[4,5,6]])
print("shape = ",arr.shape)
print("no of dimensions = ",arr.ndim)
print("no of elements ",arr.size)
print("data type of array = ",arr.dtype)
print("memory used by array in bytes = ",arr.itemsize)
```

```python
#array vectorized operations, element to elemnt operation & return will be array as well
arr1=np.array([1,2,3,4,5])
arr2=np.array([6,7,8,9,10])
print("Addition = ",arr1+arr2)
print("Substraction = ",arr2-arr1)
print("Multiplication = ",arr1*arr2)
print("division = ",arr2/arr1)
```

```python
#universal function - applies to all elemnts if array
print("square root = ",np.sqrt(arr1))
print("exponentinal = ",np.exp(arr1))
print("sine = ",np.sin(arr1))
print("natural log = ",np.log(arr1))
```

```python
#statistical concept - normalization
#normalization is nothing but mean of data should be 0 and standard deviation should be 1
data=np.array([1,2,3,4,5])
mean=np.mean(data)
std_dev=np.std(data)
print("mean = ",mean," standard deviation = ",std_dev)
#normalize data
normalized_data=(data-mean)/std_dev
print(normalized_data)
print("mean of normalized data = ",np.mean(normalized_data)," and std_dev = ",np.std(normalized_data))
```

```python
#other stats operation
median=np.median(data)
print("median = ",median)
variance=np.var(data)
print("variance = ",variance)
```
