```python
import numpy as np
import pandas as pd
```

```python
#set the random seed for repoducability
np.random.seed(50)
```

```python
n_samples=1000
class_0_ratio=0.9 #creating imbalanced dataset of 900,100
n_class_0=int(n_samples*class_0_ratio)
n_class_1=n_samples-n_class_0
n_class_0,n_class_1
```

```python
''' loc is digit near to which mean is expected, scale is S.D to 1 and size of column & normal is have normal dist to achive bell curve
'''
class_0=pd.DataFrame({
    'feature_1':np.random.normal(loc=0,scale=1,size=n_class_0),
    'feature_2':np.random.normal(loc=2,scale=1,size=n_class_0),
    'target':[0] * n_class_0
})
class_0
```

```python
class_1=pd.DataFrame({
    'feature_1':np.random.normal(loc=2,scale=1,size=n_class_1),
    'feature_2':np.random.normal(loc=2,scale=1,size=n_class_1),
    'target':[1]*n_class_1
})
```

```python
df=pd.concat([class_0,class_1]).reset_index(drop=True)
df
```

```python
df['target'].value_counts()
```

```python
#upsample:
df_minority=df[df['target'] == 1]
df_majority=df[df['target'] == 0]

from sklearn.utils import resample
df_minority_unsampled=resample(df_minority,replace=True,n_samples=len(df_majority),random_state=34)
df_minority_unsampled.shape
```

```python
df_unsampled=pd.concat([df_majority,df_minority_unsampled])
df_unsampled['target'].value_counts()
```

```python
#downsampling
class_0=pd.DataFrame({
    'feature_1':np.random.normal(loc=0,scale=1,size=n_class_0),
    'feature_2':np.random.normal(loc=2,scale=1,size=n_class_0),
    'target':[0] * n_class_0
})

class_1=pd.DataFrame({
    'feature_1':np.random.normal(loc=2,scale=1,size=n_class_1),
    'feature_2':np.random.normal(loc=2,scale=1,size=n_class_1),
    'target':[1]*n_class_1
})
df=pd.concat([class_0,class_1]).reset_index(drop=True)
```

```python
df_majority_unsampled=resample(df_majority,replace=True,n_samples=len(df_minority),random_state=42)
df_majority_unsampled.shape
```

```python
df=pd.concat([df_minority,df_majority_unsampled])
df['target'].value_counts()
```
