```python
import pandas as pd
data=[1,2,3,4,5]
series=pd.Series(data)
print(series)
print(series.dtype)
```

```python
data={'a':1,'b':2,'c':3}
series_dict=pd.Series(data)
print(series_dict)
```

```python
#manually specify data and index
data=[10,20,30]
index=['a','b','c']
pd.Series(data,index=index)
```

```python
#dataframe : 2D & heterogenous
data={
    'Name': ['Pinal','John','Jack'],
    'Age':[25,30,46],
    'City': ['nashik','pune','mumbai']
}

df=pd.DataFrame(data)
df
```
