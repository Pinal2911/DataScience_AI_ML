```python
'''### Assignment 1: Reading a File

Write a function that reads the contents of a file named `sample.txt` and prints each line.'''

with open('sample.txt','r') as file:
    print(file.read())
```

```python
'''### Assignment 2: Writing to a File

Write a function that writes a list of strings to a file named `output.txt`, with each string on a new line.'''

str_list=["hello","world","welcome","bye"]
with open('output.txt','w') as file:
    for i in str_list:
        file.write(i+"\n")

with open('output.txt','r') as file:
    print(file.read())
```

```python
'''### Assignment 3: Copying a File

Write a function that copies the contents of a file named `source.txt` to a new file named `destination.txt`.'''
with open('source.txt','r') as f:
    content=f.read()
    
with open('dest.txt','w+') as f:
    f.write(content)
    f.seek(0)
    print(f.read())
```

```python
'''### Assignment 5: Counting Words in a File

Write a function that reads the contents of a file named `document.txt` and returns the number of words in the file.'''
def count_words():
    with open('log.txt','r') as f:
        data=f.read()
        splited_data=str(data).split(' ')
        print(splited_data)
        print("total words ",len(splited_data))
count_words()
```

```python
'''### Assignment 6: Finding and Replacing Text

Write a function that finds and replaces all occurrences of a given word in a file named `data.txt` with another word.'''
def replace_word():
    with open('log.txt','r+') as f:
        data=f.read()
        data=str(data)
        data=data.replace("data.txt","replaced")
        f.seek(0)
        f.write(data)
        f.seek(0)
        print(f.read())
replace_word()
```

```python
'''### Assignment 7: Reading a File in Reverse

Write a function that reads the contents of a file named `reverse.txt` and prints each line in reverse order.'''

with open('log.txt','r') as f:
    content=f.readlines()
    for c in content:
        print(c[::-1])
```

```python
'''### Assignment 8: Counting Lines, Words, and Characters

Write a function that reads the contents of a file named `stats.txt` and returns the number of lines, words, and characters in the file.'''

with open('log.txt','r') as f:
    content=f.read()
    print(content)
    print("no of lines = ",len(str(content).split('\n')))
    print("no of words = ",len(str(content).split(' ')))
    print("no of character = ",len(str(content).replace(' ','')))

```

```python
'''### Assignment 12: Binary File Operations

Write a function that reads a binary file named `image.bin` and writes its contents to another binary file named `copy_image.bin`.'''

with open('image.bin','rb') as f:
    bin_c=f.read()
with open('copy_image.bin','wb') as f:
    f.write(bin_c)
    
```
