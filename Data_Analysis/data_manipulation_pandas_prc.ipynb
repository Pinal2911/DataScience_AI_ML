```python
#fill null values with mean
df['Sales_fillNA']=df['Sales'].fillna(df['Sales'].mean())
```

```python
#datatypes of column
df.dtypes
```

```python
#column rename
df.rename(columns={'Date':'Sales Date'})
df.head()
```

```python
#chnage datatype of column
df['Value_New_dt']=df["Value"].fillna(df['Value'].mean()).astype(int)
```

```python
df.dtypes
```

```python
#function apply on any dataframe
df['function_value']=df['Value_New_dt'].apply(lambda x:x*2)
```

```python
df.head()
```

```python
#data aggregation and grouping
grouped_mean=df.groupby('Product')['Value'].mean()
grouped_mean
```

```python
#sum based on 2 categorical columns
grouped_sum=df.groupby(['Product','Region'])['Value'].sum()
grouped_sum
```

```python
#multiple agg function
grouped_agg=df.groupby('Product')['Value'].agg(['sum','mean','count'])
grouped_agg
```

```python
#merge and join dataframes
df1=pd.DataFrame({'Key':['A','B','C'],'Value':[1,2,3]})
df1
df2=pd.DataFrame({'Key':['A','B','D'],'Value':[4,5,6]})
df2
#merge dataframes on column

pd.merge(df1,df2,on='Key',how='inner')
```

```python
pd.merge(df1,df2,on='Key',how='outer')
```

```python
pd.merge(df1,df2,on='Key',how='left')
```

```python
pd.merge(df1,df2,on='Key',how='right')
```
