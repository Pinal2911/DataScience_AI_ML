```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
df=pd.DataFrame({'color':
['red','blue','green','green','red','blue'] 
})
```

```python
df.head()
```

```python
#instance of encoder
#it will provide array directly no need to do to_array sparse_array=False
encoder=OneHotEncoder(sparse_output=False)
```

```python
#perform fit & tranform , fit -> learns from categorical data & transform does transformation from categorical to data
encoded=encoder.fit_transform(df[['color']])
```

```python
encoder_df=pd.DataFrame(encoded,columns=encoder.get_feature_names_out())
```

```python
encoder_df
```

```python
pd.concat([df,encoder_df],axis=1)
```
