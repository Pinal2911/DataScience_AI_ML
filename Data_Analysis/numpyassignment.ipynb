```python
'''1. Create a NumPy array of shape (5, 5) filled with random integers. Compute the mean, median, standard deviation, and variance of the array.'''

a4=np.random.randint(1,100,size=(5,5))
print(a4)

print("mean  = ",np.mean(a4))
print("median = ",np.median(a4))
print("stand deviation = ",np.std(a4))
print("variance = ",np.var(a4))
```

```python
'''2. Create a NumPy array of shape (3, 3) with values from 1 to 9. Normalize the array (i.e., scale the values to have a mean of 0 and a standard deviation of 1).'''

#data normalization

a5=np.random.randint(1,10,size=(3,3))
print("mean = ",np.mean(a5))
print("sd = ",np.std(a5))
#normalization=(data - mean)/std_dev
normalized_array=a5-np.mean(a5)/np.std(a5)
print("after normalization array = ")
print(normalized_array)
print("after mean =",np.mean(normalized_array))
print("after std = ",np.std(normalized_array))

```

```python
'''1. Create a NumPy array of shape (3, 3) filled with random integers. Add a 1D array of shape (3,) to each row of the 2D array using broadcasting.'''

a6=np.random.randint(1,100,size=(3,3))
print("before = ",a6)
sample_add=[12,12,12]
print("after = ",a6+sample_add)
sample_sub=[20,20,20]
print("after sub= ",a6-sample_sub)
```

```python
'''1. Create a NumPy array of shape (3, 3) representing a matrix. Compute its determinant, inverse, and eigenvalues.'''
print(a6)
print("determinant = ",np.linalg.det(a6))

#since det != 0 hence proceed with iverse
print("inverse. = ",np.linalg.inv(a6))

#check if it returns back to oiginal space if we get indentity matrix back
print("a6 @ a6_inv =\n ",a6 @ np.linalg.inv(a6))

#here note diaglonal must be 1 then it is identity matrix, ignore others
#check if it identity matrix

print("identity matrix verification = ",np.allclose(a6 @ np.linalg.inv(a6), np.eye(3)))

eigvalues,eigvector=np.linalg.eig(a6)

print("Eigne Value = ",eigvalues)
print("Eigen Vector = ")
print(eigvector)
```
