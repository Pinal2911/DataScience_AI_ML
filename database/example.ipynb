```python
import sqlite3
conn=sqlite3.connect('example.db')
cursor=conn.cursor()
cursor.execute('''
CREATE TABLE IF NOT EXISTS EMP(
               ID INTEGER PRIMARY KEY,
               NAME VARCHAR NOT NULL,
               AGE INTEGER,
               DEPT TEXT)
               ''')
conn.commit()
```

```python
#insert data in emp
cursor.execute('''
INSERT INTO EMP(NAME,AGE,DEPT) VALUES ('P',24,'DS')
               ''')
conn.commit()
```

```python
#select / fetch the data
cursor.execute('SELECT * FROM EMP')
rows=cursor.fetchall()
for r in rows:
    print(r)
```

```python
#upadte the database
cursor.execute('''
update emp set age=23 where name='P'
''')
conn.commit()

cursor.execute('SELECT * FROM EMP')
rows=cursor.fetchall()
for r in rows:
    print(r)
```

```python
#delete from table
cursor.execute('''
delete from emp where name='P'
''')
conn.commit()
```
