```python
def even(n):
    if n%2==0:
        return True
```

```python
list(filter(even,[1,2,3,4,5,6]))
```

```python
#filter with lambda
greater_than=list(filter(lambda x:x>5,[12,3,4,5,6,]))
greater_than
```

```python
#multiple con
list(filter(lambda x:x>5 and x%2 ==0,[1,2,3,4,5,6,7,8]))
```

```python
#filter with dict
people=[
    {'name':'Pinal','age':24},
    {'name':'Ruchit','age':27}
]

def age_filter(person):
    return person['age']>25

list(filter(age_filter,people))
```
