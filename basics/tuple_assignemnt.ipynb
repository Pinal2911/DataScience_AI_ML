```python
l1=[1,2,3,4,5]
t5=tuple(l1)
t5
```

```python
ll2=list(t5)
ll2.append(7)
t5=tuple(ll2)
t5
```

```python
def calculate(t5):
    min_val=5000
    max_val=0
    sum=0
    i=0
    for i in t5:
        min_val=min(i,min_val)
        max_val=max(i,max_val)
        sum+=i
    return min_val,max_val,sum
```

```python
print(calculate(t5))
```
